## 8. Advanced usage examples

```python
# Batch query processing
batch_questions = [
    "หลักสูตรเกษตรศาสตร์คืออะไร",
    "การสมัครเรียนต้องทำอย่างไร",
    "มีวิชาอะไรบ้างในหลักสูตร"
]

batch_results = app.rag_pipeline.batch_query(batch_questions)

print("📋 Batch Query Results:")
for i, result in enumerate(batch_results, 1):
    print(f"\n{i}. Q: {result['question']}")
    print(f"   A: {result['answer']}")
    print(f"   Status: {result['status']}")
```

```python
# Search for similar documents
search_query = "หลักสูตรเกษตร"
similar_docs = app.rag_pipeline.get_similar_documents(search_query, k=3)

print(f"🔍 Similar documents for: '{search_query}'")
for i, doc_info in enumerate(similar_docs, 1):
    print(f"\n{i}. Source: {doc_info['metadata'].get('source', 'Unknown')}")
    print(f"   Category: {doc_info['metadata'].get('category', 'Unknown')}")
    print(f"   Preview: {doc_info['content_preview']}")
```

## Summary of Improvements

The refactored code provides:

### 🏗️ **Modular Architecture**
- **`config.py`**: Centralized configuration management
- **`text_processing.py`**: Thai text processing utilities
- **`document_processor.py`**: Document loading and processing
- **`vector_store.py`**: ChromaDB operations
- **`rag_pipeline.py`**: Query processing and response generation
- **`web_interface.py`**: Gradio UI components
- **`main.py`**: Application orchestration

### 🚀 **Key Benefits**
1. **Maintainability**: Each component has a single responsibility
2. **Testability**: Individual modules can be tested separately
3. **Reusability**: Components can be used independently
4. **Configuration**: Centralized, flexible configuration system
5. **Error Handling**: Better error management and logging
6. **Documentation**: Clear docstrings and type hints
7. **Scalability**: Easy to extend and modify individual components

### 🛠️ **Usage Patterns**
- **Quick Start**: `AgroRAGApplication().run_full_pipeline(folders, urls)`
- **Step-by-step**: Individual component control
- **Programmatic**: Batch processing and API usage
- **Interactive**: Advanced web interface with multiple tabs
